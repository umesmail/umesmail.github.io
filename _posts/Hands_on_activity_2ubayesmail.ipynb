```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)

y = x**2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Plot of y = x^2")
plt.show()
```

```python
y1 = x**2
y2 = 2*x + 5

plt.plot(x, y1, color="blue", label="y = x^2")
plt.plot(x, y2, color="red", label="y = 2x + 5")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Plot of Two Functions")
plt.legend()
plt.show()
```

```python
import pandas as pd

df = pd.read_excel("chrosome_basepairs.xlsx")

df
```

```python
total_basepairs = df["basepais(bp)"].sum()

length_angstrom = total_basepairs * 3.4
length_meters = length_angstrom * 1e-10

print("Total base pairs:", total_basepairs)
print("Genome length in Angstroms:", length_angstrom)
print("Genome length in meters:", length_meters)
```

```python
total_mbp = total_basepairs / 1000000

years = [2001, 2010, 2011, 2020]
cost_per_mbp = [10000, 1, 0.10, 0.01]

total_costs = []

for cost in cost_per_mbp:
    total_costs.append(total_mbp * cost)

print("Genome size in Mbp:", total_mbp)
print("Total sequencing costs:", total_costs)
```

```python
import matplotlib.pyplot as plt

plt.plot(years, total_costs, marker="o")

plt.xlabel("Year")
plt.ylabel("Total Cost ($)")
plt.title("Cost to Sequence the Whole Genome")
plt.yscale("log")
plt.xticks(years)
plt.grid()

plt.show()
```
